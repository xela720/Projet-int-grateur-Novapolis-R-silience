Ce petit réseau est synthétique. Le trajet A → C → D → E est choisi à la main pour montrer le mouvement : aucun algorithme ne le calcule ici. Cette démonstration visuelle ne constitue pas encore le projet Novapolis. Le module de graphe devra être adapté à l'énoncé.

```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
from graphe import creer_exemple

graphe = creer_exemple()
positions = graphe.positions()
chemin = ["A", "C", "D", "E"]

# shortcut: trajet choisi à la main, à remplacer par le résultat d'un algorithme.
points = []
for depart, arrivee in zip(chemin, chemin[1:]):
    assert arrivee in graphe.voisins(depart), "Le trajet contient une route absente."
    x1, y1 = positions[depart]
    x2, y2 = positions[arrivee]
    for etape in range(20):
        progression = etape / 20
        points.append((x1 + progression * (x2 - x1),
                       y1 + progression * (y2 - y1)))
points.append(positions[chemin[-1]])

assert points[0] == positions[chemin[0]]
assert points[-1] == positions[chemin[-1]]
```

```python
with plt.style.context("dark_background"):
    fig, ax = plt.subplots(figsize=(8, 4.5), dpi=85)

for depart, arrivee, poids in graphe.aretes():
    x1, y1 = positions[depart]
    x2, y2 = positions[arrivee]
    ax.plot([x1, x2], [y1, y2], color="#596579", linewidth=2, zorder=1)
    ax.text((x1 + x2) / 2 + 0.10, (y1 + y2) / 2 + 0.10,
            str(poids), color="#cbd5e1", fontsize=11)

for nom, (x, y) in positions.items():
    ax.scatter(x, y, s=130, color="#38bdf8", zorder=3)
    ax.text(x, y - 0.30, nom, color="white", fontsize=13, ha="center")

trace, = ax.plot([], [], color="#4ade80", linewidth=4, zorder=2)
point, = ax.plot([], [], "o", color="#fbbf24", markersize=12, zorder=4)
ax.set_title("Trajet choisi à la main : A → C → D → E", color="white", pad=20)
ax.set_xlim(-0.7, 4.7)
ax.set_ylim(-0.7, 2.7)
ax.set_aspect("equal")
ax.axis("off")
fig.tight_layout()


def actualiser(numero):
    portion = points[:numero + 1]
    trace.set_data([p[0] for p in portion], [p[1] for p in portion])
    x, y = points[numero]
    point.set_data([x], [y])
    return trace, point


for numero in [0, len(points) // 2, len(points) - 1]:
    actualiser(numero)
    assert (point.get_xdata()[0], point.get_ydata()[0]) == points[numero]
actualiser(0)

animation = FuncAnimation(fig, actualiser, frames=len(points), interval=60,
                          blit=True, repeat=True)
plt.close(fig)
display(HTML(animation.to_jshtml(default_mode="loop")))
```

Le déplacement prend environ 3,6 secondes. Pour le ralentir, remplace `interval=60` par `interval=120` et réexécute la cellule du dessin.

La vitesse est illustrative : elle ne représente pas les temps de trajet du projet. L'affichage et l'animation resteront séparés du chronométrage des futurs algorithmes.
